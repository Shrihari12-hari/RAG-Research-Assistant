```python

from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
```

```python

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

print("Text splitter configured successfully!")
print("Chunk size: 1000")
print("Chunk overlap: 150")
```

```python

documents = []

for page in all_pages:

    if page["text"]:

        doc = Document(
            page_content=page["text"],
            metadata={
                "paper_id": page["paper_id"],
                "paper_title": page["paper_title"],
                "file_name": page["file_name"],
                "page_number": page["page_number"]
            }
        )

        documents.append(doc)

print("Total page documents:", len(documents))
```

```python

chunks = text_splitter.split_documents(documents)

# Assign a unique chunk ID
for index, chunk in enumerate(chunks):
    chunk.metadata["chunk_id"] = f"CHUNK_{index + 1:04d}"

print("Total chunks generated:", len(chunks))

# Chunking strategy justification
print(
    "\nChunking strategy: Page-wise recursive splitting "
    "preserves page citations while respecting paragraph boundaries."
)

# PDF layout limitation
print(
    "Note: Two-column layouts, equations and tables may affect "
    "extraction and chunk quality."
)
```

```python

from collections import Counter

chunk_counts = Counter(
    chunk.metadata["paper_id"] for chunk in chunks
)

for paper_id, count in chunk_counts.items():

    title = PAPER_INFO[
        next(
            filename for filename, info in PAPER_INFO.items()
            if info["paper_id"] == paper_id
        )
    ]["paper_title"]

    print(f"\n{paper_id}: {title}")
    print("Total chunks:", count)

print("\nTOTAL CHUNKS:", len(chunks))
```

```python
import os
import requests
from dotenv import load_dotenv
from langchain_core.embeddings import Embeddings

load_dotenv()

jina_api_key = os.getenv("JINA_API_KEY")

if jina_api_key:
    print("Jina API key loaded successfully!")
else:
    print("Jina API key not found!")
```

```python
class JinaAPIEmbeddings(Embeddings):

    def __init__(self, api_key):
        self.api_key = api_key
        self.url = "https://api.jina.ai/v1/embeddings"
        self.model = "jina-embeddings-v3"

    def embed_documents(self, texts):

        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json"
        }

        embeddings = []

        for i in range(0, len(texts), 32):

            batch = texts[i:i + 32]

            payload = {
                "model": self.model,
                "task": "retrieval.passage",
                "input": batch
            }

            response = requests.post(
                self.url,
                headers=headers,
                json=payload,
                timeout=120
            )

            response.raise_for_status()

            data = sorted(
                response.json()["data"],
                key=lambda item: item["index"]
            )

            embeddings.extend(
                item["embedding"] for item in data
            )

        return embeddings

    def embed_query(self, text):

        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json"
        }

        payload = {
            "model": self.model,
            "task": "retrieval.query",
            "input": [text]
        }

        response = requests.post(
            self.url,
            headers=headers,
            json=payload,
            timeout=120
        )

        response.raise_for_status()

        return response.json()["data"][0]["embedding"]
```

```python
embedding_model = JinaAPIEmbeddings(jina_api_key)

print("Jina embedding model initialized successfully!")
```

```python
import chromadb

chroma_client = chromadb.PersistentClient(
    path=str(BASE_DIR / "vectorstore")
)

collection = chroma_client.get_or_create_collection(
    name="rag_research_papers",
    metadata={"hnsw:space": "cosine"}
)

print("ChromaDB initialized successfully!")
print("Collection:", collection.name)
```

```python
collection.upsert(
    ids=[
        chunk.metadata["chunk_id"]
        for chunk in chunks
    ],
    documents=[
        chunk.page_content
        for chunk in chunks
    ],
    metadatas=[
        chunk.metadata
        for chunk in chunks
    ],
    embeddings=all_embeddings
)

print("All research chunks stored successfully!")
print("Total records:", collection.count())
```

```python
query = "What is Retrieval Augmented Generation and how does it work?"

query_embedding = embedding_model.embed_query(query)

search_results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

for i in range(3):

    print("\n" + "=" * 60)
    print(f"RESULT {i + 1}")

    print("Paper:", search_results["metadatas"][0][i]["paper_title"])
    print("Page:", search_results["metadatas"][0][i]["page_number"])
    print("Chunk ID:", search_results["ids"][0][i])

    print("\nRetrieved Text:")
    print(search_results["documents"][0][i][:500])
```

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os

load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")

if groq_api_key:
    print("Groq API key loaded successfully!")
else:
    print("Groq API key not found!")
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    api_key=groq_api_key
)

print("Groq LLM initialized successfully!")
```

```python
response = llm.invoke(
    "Explain Retrieval Augmented Generation in simple words."
)

print(response.content)
```

```python
def ask_research_assistant(question):

    # Step 1: Convert question into embedding
    query_embedding = embedding_model.embed_query(question)

    # Step 2: Retrieve relevant chunks
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=3
    )

    # Step 3: Prepare context
    context = ""

    for i in range(len(results["documents"][0])):

        metadata = results["metadatas"][0][i]
        document = results["documents"][0][i]

        context += f"""
        Paper: {metadata['paper_title']}
        Page: {metadata['page_number']}
        Content: {document}
        """

    # Step 4: Create prompt
    prompt = f"""
    You are an AI research assistant specializing in
    Machine Learning and Deep Learning.

    Answer the question using only the provided research-paper context.

    If the answer is not available in the context,
    clearly say that the provided papers do not contain
    enough information.

    Research Context:
    {context}

    Question:
    {question}

    Provide a clear and simple explanation.
    """

    # Step 5: Generate answer using Groq
    response = llm.invoke(prompt)

    # Step 6: Display answer and sources
    print("RESEARCH ASSISTANT ANSWER")
    print("=" * 60)
    print(response.content)

    print("\nSOURCES")
    print("=" * 60)

    for i, metadata in enumerate(results["metadatas"][0]):

        print(f"{i + 1}. {metadata['paper_title']}")
        print(f"   Page: {metadata['page_number']}")
        print(f"   Chunk: {results['ids'][0][i]}")
```

```python
question = "What is Retrieval Augmented Generation and how does it work?"

ask_research_assistant(question)
```

```python
question = "What is parametric memory? in short"

ask_research_assistant(question)
```
